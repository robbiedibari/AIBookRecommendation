```python
# Install dependencies as needed:
# pip install kagglehub[hf-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter

# Download the latest version
path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files: ", path)
```

```python
import pandas as pd 
books = pd.read_csv(f'{path}/books.csv')
```

```python
# First step inspect the data 
import seaborn as sns 
import matplotlib.pyplot as plt 
ax = plt.axes()
sns.heatmap(books.isna().transpose(), cbar = False, ax=ax)

plt.xlabel('Columns')
plt.ylabel("Missing values")
plt.show()
```

```python
import numpy as np 
books['missing_description'] = np.where(books['description'].isna(),1,0)
books['age_of_book'] = 2025- books['published_year']
```

```python
# Reasoning, verify correlation between missing_description and other variables 
# close to 1 or -1 strong correlation 
# close 0 no correlation 

columns_of_interest = ['num_pages',"age_of_book", "missing_description", "average_rating"]
correlation_matrix = books[columns_of_interest].corr(method="spearman")

sns.set_theme(style="white")
plt.figure(figsize= (8,6))

heatmap = sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap ='coolwarm',
                      cbar_kws={"label": "Spearman correlation" })
heatmap.set_title("Correlation Heatmap")
plt.show()
```

```python
# Delete observation with missing values: 
book_missing = books[~(books["description"].isna()) &
      ~ (books["num_pages"].isna()) &
      ~ (books["average_rating"].isna())&
      ~ (books["published_year"].isna())
]
# ~ this is the negation sign 
```

```python
# new data frame which has only those field  
book_missing
```

```python
# distribution of categoires 
book_missing['categories'].value_counts().reset_index().sort_values("count", ascending= False)
```

```python
book_missing['words_in_description'] = book_missing["description"].str.split().str.len()
```

```python
# cleaning less than 25 words in descirptions 
book_missing_25_words = book_missing[book_missing["words_in_description"] >= 25]
book_missing_25_words
```

```python
# Create new_filed title_subtitle  when subtitle is missing just use the title when there is a sub title aggregate the title and subtitle 
book_missing_25_words['title_and_subtitle'] = (
    np.where(
    book_missing_25_words['subtitle'].isna(), book_missing_25_words['title'],
    book_missing_25_words[['title', 'subtitle']].astype(str).agg(": ".join, axis=1))
)
book_missing_25_words
```

```python
# new description column with unique identifiers 
book_missing_25_words["tagged_description"] = book_missing_25_words[["isbn13", "description"]].astype(str).agg(" ".join, axis=1)
book_missing_25_words
```

```python
(
    book_missing_25_words.drop(["subtitle", "missing_description", "age_of_book", "words_in_description"], axis = 1)
    .to_csv('book_cleaned.csv', index = False)
)
```
